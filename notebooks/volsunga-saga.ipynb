# Völsunga saga

The *Völsunga saga* is read from [norsecorpus](https://github.com/clemsciences/old_norse_corpus), whose TEI text is already split into chapters, paragraphs, sentences and words.

```python
import norsecorpus.reader as ncr
```

```python
available_texts = ncr.get_available_texts()
list(available_texts)
```

```python
volsunga_saga = ncr.read_tei_words(available_texts["volsunga.xml"])
```

```python
print(f"There are {len(volsunga_saga)} chapters.")
```

Each chapter becomes a list of sentences, each sentence being a list of words.

```python
tokenized_chapters = [[sentence for paragraph in chapter for sentence in paragraph] for chapter in volsunga_saga]
```
