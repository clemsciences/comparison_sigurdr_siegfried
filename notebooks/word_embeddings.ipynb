# Text comparison

```python
import norsecorpus.reader as ncr
from gensim.models import Word2Vec
```

```python
available_texts = ncr.get_available_texts()
volsunga_saga = ncr.read_tei_words(available_texts["volsunga.xml"])
```

```python
sentences = [sentence for chapter in volsunga_saga for paragraph in chapter for sentence in paragraph]
print(sentences[0])
```

The saga has only about 31,000 words: with `min_count=1`, most words are seen a few times, so the similarities below are not reliable.

```python
model = Word2Vec(sentences, min_count=1, vector_size=50, seed=1, workers=1)
```

```python
len(model.wv)
```

```python
model.wv["ok"]
```

```python
model.wv.similarity("konu", "manni")
```
